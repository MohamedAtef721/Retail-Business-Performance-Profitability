```python
# 1. Missing values
print("Missing values by column:")
print(df.isna().sum().sort_values(ascending=False))

# 2. Future dates
today = pd.Timestamp.today().normalize()

print("\nToday's date:", today.date())
print("Future order dates:", (df["Order Date"] > today).sum())
print("Future ship dates:", (df["Ship Date"] > today).sum())

# 3. Duplicate order-product combinations
print(
    "\nDuplicate Order ID + Product ID combinations:",
    df.duplicated(subset=["Order ID", "Product ID"]).sum()
)

# 4. Distinct values in important categorical columns
for col in ["Region", "Segment", "Category", "Ship Mode"]:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```

```python
# 1. Future dates
today = pd.Timestamp("2026-10-24")

print("Future Order Dates:",
      (df["Order Date"] > today).sum())

print("Future Ship Dates:",
      (df["Ship Date"] > today).sum())

# 2. Duplicate order-product combinations
print(
    "Duplicate Order ID + Product ID combinations:",
    df.duplicated(
        subset=["Order ID", "Product ID"]
    ).sum()
)

# 3. Full categorical distributions
for col in ["Region", "Segment", "Category", "Ship Mode"]:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).to_string())
```

```python
# Fixed analysis date
today = pd.Timestamp("2026-10-24")

# 1. Inspect future-dated records
future_orders = df[df["Order Date"] > today]

print("Future order date range:")
print(future_orders["Order Date"].min())
print(future_orders["Order Date"].max())

print("\nDistinct future Order IDs:")
print(future_orders["Order ID"].nunique())

print("\nSample future-dated records:")
print(
    future_orders[
        ["Order ID", "Order Date", "Ship Date",
         "Product ID", "Sales", "Profit"]
    ].head(10).to_string(index=False)
)

# 2. Inspect all repeated Order ID + Product ID pairs
repeated_pairs = df[
    df.duplicated(
        subset=["Order ID", "Product ID"],
        keep=False
    )
].sort_values(["Order ID", "Product ID"])

print("\nAll repeated Order ID + Product ID records:")
print(
    repeated_pairs[
        ["Order ID", "Product ID", "Order Date",
         "Sales", "Quantity", "Discount", "Profit"]
    ].to_string(index=False)
)
```

```python
# Group repeated order-product combinations
repeated_summary = (
    df.groupby(["Order ID", "Product ID"], as_index=False)
      .agg(
          Row_Count=("Row ID", "count"),
          Total_Sales=("Sales", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Total_Profit=("Profit", "sum"),
          Distinct_Discounts=("Discount", "nunique")
      )
      .query("Row_Count > 1")
      .sort_values("Row_Count", ascending=False)
)

print("Repeated combinations:", len(repeated_summary))
print(repeated_summary.to_string(index=False))
```
