```python
print("Exact duplicate rows:", data.duplicated().sum())

print("Duplicate Row IDs:", data["Row ID"].duplicated().sum())

print("Unique Order IDs:", data["Order ID"].nunique())

print("Unique Product IDs:", data["Product ID"].nunique())
```

```python
print("Sales <= 0:", (data["Sales"] <= 0).sum())
print("Quantity <= 0:", (data["Quantity"] <= 0).sum())
print("Profit < 0:", (data["Profit"] < 0).sum())
print("Discount < 0:", (data["Discount"] < 0).sum())
print("Discount > 1:", (data["Discount"] > 1).sum())

print("Ship Date before Order Date:", (data["Ship Date"] < data["Order Date"]).sum())

print("\nFirst Order Date:", data["Order Date"].min())
print("Last Order Date:", data["Order Date"].max())
```
